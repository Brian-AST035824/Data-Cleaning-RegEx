The pattern has three main sections: (^[\w]*) — First capturing group - ( and ) create a capturing group. - ^ means the beginning of the string. - [\w] matches a word character, such as a letter, digit, or underscore. - * means zero or more occurrences.
Meaning: Capture the word characters at the beginning of the name.

(?:.*) — Non-capturing group - (?:...) groups a pattern without creating an output column. - . matches any character except a newline by default. - * means zero or more characters.
Meaning: Match the characters between the first and last parts without extracting them as a separate column.
([\w]*$) — Second capturing group - () creates another capturing group. - [\w]* matches zero or more word characters. - $ means the end of the string. Meaning: Capture the word characters at the end of the name.

Let's understand the advanced features here:
- (?P<Years>...) creates a named capturing group called Years.
- \d+ captures one or more digits.
- \s+ matches whitespace.
- , matches the comma separating years and days.
- \s* allows zero or more spaces after the comma.
- (?P<Days>...) creates another named capturing group called Days.
- ? after s allows either singular or plural, such as year or years.

Here:
- pd.to_numeric() converts extracted text into numeric values.
- errors='coerce' converts unrecognizable values to missing values rather than raising an error.
- days / 365 converts the extra days into a fraction of a year.
- .round(2) rounds the result to two decimal places.

```python
df['Age_Start_Original'] = df['Age_Start_Original'] if 'Age_Start_Original' in df.columns else df['Age_Start'].copy()

years = pd.to_numeric(df['Age_Start'].astype('string').str.extract(r'(\d+)\s+years?')[0])
days = pd.to_numeric(df['Age_Start'].astype('string').str.extract(r',\s*(\d+)\s+days?')[0])

# Calculate age in decimal years to 2 decimal places
df['Age_Start'] = ((years + days / 365).round(2).map(
        lambda x: f'{x:.2f} years' if pd.notna(x) else pd.NA))

# Extract the starting date in the format Mon DD, YYYY
df['Date_Start'] = df['Age_Start_Original'].astype('string').str.extract(
    r'([A-Z][a-z]{2} \d{1,2}, \d{4})')
# df['Date_Start'] = pd.to_datetime(df['Date_Start']) # Convert date to datetime
df.head()
```

Understanding the date pattern
r'([A-Z][a-z]{2} \d{1,2}, \d{4})'
- [A-Z] — one uppercase letter, such as J.
- [a-z]{2} — two lowercase letters, completing a three-letter month such as Jan.
- \d{1,2} — one or two digits for the day.
- ,  — a comma followed by a space.
- \d{4} — four digits for the year.
- Parentheses () capture the complete date.

```python
years = pd.to_numeric(df['Age_End'].astype('string').str.extract(r'(\d+)\s+years?')[0])
days = pd.to_numeric(df['Age_End'].astype('string').str.extract(r',\s*(\d+)\s+days?')[0])# Convert extracted values to numeric

# Calculate age in decimal years to 2 decimal places
df['Age_end'] = ((years + days / 365).round(2).map(
        lambda x: f'{x:.2f} years' if pd.notna(x) else pd.NA))
# Extract the starting date in the format Mon DD, YYYY
df['Date_End'] = df['Age_End'].astype('string').str.extract(
    r'([A-Z][a-z]{2} \d{1,2}, \d{4})')
df 
```

```python
df.drop(columns=['Age_End', 'Age_Start_Original'], inplace=True) # 1. Drop the unwanted columns

# 2. Specify the columns that should appear first
first_columns = ['President','Born','Age_Start','Date_Start','Age_end','Date_End']

# 3. Append all remaining columns in their existing order
remaining_columns = [
    col for col in df.columns
    if col not in first_columns]
df = df[first_columns + remaining_columns]
df.head()
```

Understanding the regular expressions
- r'\[[a-zA-Z]\]' matches a letter enclosed in square brackets, such as [c] or [d].
- .str.replace(..., '', regex=True) replaces those markers with an empty string.
- .str.strip() removes whitespace at the beginning and end.
- r'\s+' matches one or more whitespace characters, replacing repeated spaces with a single space.

```python
# Remove citation markers such as [c] and [d]
df['Post_Presidency_Timespan'] = (df['Post_Presidency_Timespan']
    .str.replace(r'\[[a-zA-Z]\]', '', regex=True)
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True))
df['Post_Presidency_Timespan'].value_counts(dropna=False)
```

```python
# Extract years and days using regular expressions
years = pd.to_numeric(df['Post_Presidency_Timespan'].str.extract(r'(?:(\d+)\s+years?,?\s*)?(?:(\d+)\s+days?)?', expand=True)[0])
days = pd.to_numeric(df['Post_Presidency_Timespan'].str.extract(r'(?:(\d+)\s+years?,?\s*)?(?:(\d+)\s+days?)?', expand=True)[1])

# Round days to the nearest year only when years are also present
rounded_years = (years + (days / 365).round()).astype('Int64')

# Create the cleaned column
df['Post_Presidency'] = df['Post_Presidency_Timespan'].where(
    years.isna() | days.isna(), rounded_years.astype('string') + ' years')

df
```

```python
# Extract years and days using regular expressions
parts = df['Age_At_Demise'].str.extract(
    r'(?:(\d+)\s+years?,\s*)?(\d+)\s+days?'
)

years = pd.to_numeric(parts[0])
days = pd.to_numeric(parts[1])

# Round days to the nearest year only when years are present
rounded_years = (years + (days / 365).round()).astype('Int64')

# Create the cleaned column
df['Age_At_Demise'] = df['Age_At_Demise'].where(
    years.isna() | days.isna(),
    rounded_years.astype('string') + ' years')
# df['Died_Date'] = pd.to_datetime(df['Died_Date'])
df
```

```python
df['Died_Date'] = df['Died_Date'].str.replace(r'[()]', '', regex=True)
df['Died_Date'] = pd.to_datetime(df['Died_Date'])

df['Died_Date'] = df['Died_Date'].astype('string').fillna('living')
df
```
